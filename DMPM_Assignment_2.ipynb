```python
#WAP to print the mode of the dataitem without using a function.
li=[]
max=0
sum=0
n=int(input("how many data item:"))
for i in range(n):
  val=int(input(f"enter the number {i+1}:"))
  li.append(val)

mode = li[0]
max_count = 0
for i in li:
  count = 0
  for x in li:
      if x == i:
          count += 1
  if count > max_count:
        max_count = count
        mode = i
if max_count == 1:
    print("There is no unique mode .")
else:
    print(f"Mode = {mode}")
```
