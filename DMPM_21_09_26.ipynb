```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
df=sns.load_dataset("iris")
df["species"].unique()
```

```python
df["species"]=df["species"].map({"setosa":0,"versicolor":1,"virginica":2})
```

```python
X=df.drop("species",axis=1)
Y=df["species"]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42)
```

```python
model=DecisionTreeClassifier(min_samples_split=2,max_depth=3,max_leaf_nodes=3)
```

```python
model.fit(X_train,Y_train)
```

```python

Y_pred=model.predict(X_test)
```

```python

print("Accuracy_Score:",accuracy_score(Y_test,Y_pred))
```

```python
print("\n--- Calculating Entropy, Weighted Entropy, and Information Gain ---")

# Calculate initial entropy for the target variable (Y_train)
if len(Y_train) == 0:
    initial_entropy = 0
else:
    proportions_initial = Y_train.value_counts(normalize=True)
    initial_entropy = -np.sum(proportions_initial * np.log2(proportions_initial + 1e-10))
print(f"Initial Entropy of Y_train: {initial_entropy:.4f}")

n_samples_train = len(Y_train)
n_features_train = X_train.shape[1]

results = []

for feature_idx in range(n_features_train):
    feature_name = X_train.columns[feature_idx]
    thresholds = np.unique(X_train.iloc[:, feature_idx])

    for threshold in thresholds:
        left_indices = X_train.iloc[:, feature_idx] <= threshold
        right_indices = X_train.iloc[:, feature_idx] > threshold

        y_left, y_right = Y_train[left_indices], Y_train[right_indices]

        if len(y_left) == 0 or len(y_right) == 0:
            continue

        # Entropy for left child
        if len(y_left) == 0:
            entropy_left = 0
        else:
            proportions_left = y_left.value_counts(normalize=True)
            entropy_left = -np.sum(proportions_left * np.log2(proportions_left + 1e-10))

        # Entropy for right child
        if len(y_right) == 0:
            entropy_right = 0
        else:
            proportions_right = y_right.value_counts(normalize=True)
            entropy_right = -np.sum(proportions_right * np.log2(proportions_right + 1e-10))

        weight_left = len(y_left) / n_samples_train
        weight_right = len(y_right) / n_samples_train

        weighted_entropy = (weight_left * entropy_left) + (weight_right * entropy_right)
        information_gain = initial_entropy - weighted_entropy

        results.append({
            'Feature': feature_name,
            'Threshold': threshold,
            'Entropy (Left)': entropy_left,
            'Entropy (Right)': entropy_right,
            'Weighted Entropy': weighted_entropy,
            'Information Gain': information_gain
        })

# Display results in a DataFrame for better readability
results_df = pd.DataFrame(results)
display(results_df.sort_values(by='Information Gain', ascending=False).head(10))
```
