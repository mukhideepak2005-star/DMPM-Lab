```python
import pandas as pd
import numpy as np
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
df=pd.read_csv("Apriori.csv")
df
```

```python
# Preprocess the data to get transactions in the correct format
# Assuming 'Items' column contains comma-separated values, and the last item might be empty
data = df['Items'].apply(lambda x: [item.strip() for item in x.split(',') if item.strip()])

# Create a list of all unique items to use as columns for one-hot encoding
all_items = sorted(list(set(item for sublist in data for item in sublist)))

# One-hot encode the transactions
encoded_data = []
for transaction in data:
    row = [1 if item in transaction else 0 for item in all_items]
    encoded_data.append(row)

df_encoded = pd.DataFrame(encoded_data, columns=all_items)
print("One-hot Encoded DataFrame Head:")
display(df_encoded.head())
```

With the data now preprocessed, I'll apply the Apriori algorithm to find frequent itemsets with a minimum support of 60%, and then generate association rules, displaying their confidence scores.

```python
# Apply the Apriori algorithm with min_support=0.6 (60%)
frequent_itemsets = apriori(df_encoded, min_support=0.6, use_colnames=True)
print("Frequent Itemsets (min_support=60%):")
display(frequent_itemsets)

# Generate association rules using 'confidence' as the metric
# Setting a default min_threshold of 0.5 for confidence, which can be adjusted
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.5)
print("\nAssociation Rules (min_threshold=0.5 for confidence, sorted by confidence):")
display(rules.sort_values(by='confidence', ascending=False))
```

```python
df1=pd.read_csv("Chi_Sq.csv")
df1
```

```python
# Extract the contingency table data (excluding the 'Unnamed: 0' column which is the row label)
contingency_table = df1.iloc[:, 1:]

# Calculate the number of rows and columns in the contingency table
num_rows = contingency_table.shape[0]
num_cols = contingency_table.shape[1]

# Calculate degrees of freedom (df)
degrees_of_freedom = (num_rows - 1) * (num_cols - 1)

print(f"Number of rows in the contingency table: {num_rows}")
print(f"Number of columns in the contingency table: {num_cols}")
print(f"Degrees of freedom: {degrees_of_freedom}")
```

```python
# Calculate row totals
row_totals = contingency_table.sum(axis=1)

# Calculate column totals
column_totals = contingency_table.sum(axis=0)

# Calculate grand total
grand_total = contingency_table.sum().sum()

# Create an empty DataFrame for expected frequencies
expected_frequencies = pd.DataFrame(index=contingency_table.index, columns=contingency_table.columns)

# Calculate expected frequencies for each cell
for i in range(len(contingency_table.index)):
    for j in range(len(contingency_table.columns)):
        expected_frequencies.iloc[i, j] = (row_totals.iloc[i] * column_totals.iloc[j]) / grand_total

print("Observed Frequencies:")
display(contingency_table)

print("\nExpected Frequencies:")
display(expected_frequencies.round(2)) # Round for better readability
```
