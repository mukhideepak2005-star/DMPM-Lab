## Correlation coefficient and covariance analysis

### Correlation Coefficient

```python
import numpy as np
import pandas as pd
import math
```

```python
X=np.array([12,13,14,15,10,81.42,54,46,67,82,32,41,54,76,57,80])
Y=np.array([21,22,34,32,45,56,87,80,32,13,15,56,67,78,35,67])[:len(X)]
s_X=0
s_Y=0
s_XY=0
upper=0
mean_X=np.mean(X)
mean_Y=np.mean(Y)
```

```python
for i in range(len(X)):
  s_XY=s_XY+((X[i]-mean_X)*(Y[i]-mean_Y))
  s_X=s_X+((X[i]-mean_X)**2)
  s_Y=s_Y+((Y[i]-mean_Y)**2)
```

```python
Corr=s_XY/math.sqrt(s_X*s_Y)
print("Correlation coefficient is: ",Corr)
```

### Covariance matrics

```python
for j in range(len(X)):
  upper=upper+((X[j]-mean_X)*(Y[j]-mean_Y))
cov_xy=upper/(len(X)-1)
print("Covariance is: ",cov_xy)
```

### Using NumPy for Correlation and Covariance

```python
import numpy as np

# Calculate the correlation coefficient using NumPy
correlation_matrix = np.corrcoef(X, Y)
correlation_coefficient_np = correlation_matrix[0, 1]
print(f"Correlation coefficient (NumPy): {correlation_coefficient_np}")
```

```python
# Calculate the covariance matrix using NumPy
covariance_matrix_np = np.cov(X, Y, ddof=1) # ddof=1 for sample covariance
print("Covariance matrix (NumPy):\n", covariance_matrix_np)

# The covariance between X and Y is at [0, 1] or [1, 0]
cov_xy_np = covariance_matrix_np[0, 1]
print(f"Covariance of X and Y (NumPy): {cov_xy_np}")
```
