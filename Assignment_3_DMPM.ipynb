## Q_2: Plot bargraph,linegraph,piechartmhistogram,scatterplot

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x=np.array(["Apple","Banana","Mango","Papaya"])
y=np.array([10,25,5,15])
```

### a. Bargraph

```python
plt.bar(x,y)
plt.xlabel("Fruits")
plt.ylabel("Quantity")
plt.title("Bar Graph")
plt.show()
```

### b. Linegraph

```python
plt.plot(x,y)
plt.xlabel("Fruits")
plt.ylabel("Quantity")
plt.title("Line Graph")
plt.show()
```

### c. Piechart

```python
plt.pie(y,labels=x)
plt.title("Fruite Analysis ")
plt.axis("equal")
plt.show()
```

### d. Histogram

```python
data = [2, 3, 3, 4, 5, 5, 5, 6, 7, 8, 12, 14, 15, 15, 18, 20]
plt.hist(data, bins=5, edgecolor='black')
plt.title("Simple Histogram Example")
plt.xlabel("Value Range (Bins)")
plt.ylabel("Frequency")
plt.show()
```

### e. Scatterplot

```python
plt.scatter(x,y)
plt.xlabel("Fruits")
plt.ylabel("Quantity")
plt.title("Scatter Plot")
plt.show()
```

# Q-4 : Implemet the boxplot on the titanic dataset

```python
df=pd.read_csv("Titanic-Dataset.csv")
```

```python
for i in df.select_dtypes(include=[int,float]):
  plt.boxplot(df[i])
  plt.title(i)
  plt.show()
```
