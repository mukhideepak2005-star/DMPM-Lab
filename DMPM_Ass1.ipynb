```python
string=input("Enter a string:")
vcount=0
ccount=0
for i in range(len(string)):
  if string[i] in "aeiouAEIOU":
    vcount=vcount+1
  else:
    ccount+=1
print(f"no of vowels {vcount} no of consonant {ccount}.")
```

```python
n=int(input("enter a number to check prime:"))
count=0
for i in range(1,n+1):
  if(n%i==0):
    count+=1
if(count==2):
  print("prime.")
else:
  print("not prime.")
```

```python
def fact(n):
  fact=1
  for i in range(1,n+1):
    fact=fact*i
  return fact
print("factorial of 5 is ",fact(5))
```

```python
li=[1,2,2,3,4,5,3,4,4,2]
element=int(input("enter an element:"))
freq=0
for i in li:
  if(i==element):
    freq+=1
if element not in li:
  print("item not present in the list!!")
print("frequency of given item is:",freq)
```

```python
li=[1,2,3,45,65,43,7,98,30,20]
max=li[0]
min=li[0]
for i in li:
  if max<i:
    max=i
  elif(min>i):
    min=i
print(f"maximum value is {max} & minimum value is {min}.")
```
