## implementation of min_max normalization on a datasset

```python
import pandas as pd
import numpy as np

data=np.array([10,20,13,15,40,30,60,70,30,35,47,67,54,43,21,42])
min=np.min(data)
max=np.max(data)
normalized_vals=(data-min)/(max-min)
print(normalized_vals)
```

## implement the min max normalization on a dataset

```python
import pandas as pd
import numpy as np
import seaborn as sns

iris=sns.load_dataset("iris")
iris
```

```python
cols=["sepal_length","sepal_width","petal_length","petal_width"]
iris_normalized=iris.copy()
for i in cols:
  min_val = np.min(iris[i])
  max_val = np.max(iris[i])
  iris_normalized[i]=(iris[i]-min_val)/(max_val-min_val)
iris_normalized
```

## implement the decimal scaling normalization on appropriate dataset

```python
import numpy as np

data=np.array([1200,1500,1400,2340,4563,2340,1390,4500,5678,4533,2345])

# Find the maximum absolute value in the data
max_val = np.max(np.abs(data))

# Calculate j, the smallest integer such that max_val / (10**j) < 1
# This is equivalent to finding j such that 10**j >= max_val
j = 0
while 10**j <= max_val:
    j += 1

# Perform decimal scaling normalization
normalized_data = data / (10**j)

print(f"Normalized data: {normalized_data}")
```

## implement the decimal scaling normalization on appropriate dataset

```python
iris=sns.load_dataset("iris")

cols=["sepal_length","sepal_width","petal_length","petal_width"]
normalized_iris=iris.copy()
for i in cols:
  max_val=np.max(np.abs(iris[i]))
  j=0
  while 10**j<=max_val:
    j+=1
  normalized_iris[i]=iris[i]/(10**j)
normalized_iris
```

## implement the zscore normalization on 10 datapoints entered by user

```python
datapoints=[]
for i in range(11):
  point=int(input("enter the datapoint:"))
  datapoints.append(point)
datapoints=np.array(datapoints)
mean=np.mean(datapoints)
std=np.std(datapoints)
normalized_data=(datapoints-mean)/std
print(normalized_data)
```

## on a dataset find the normalized values using z-score normalization

```python
iris=sns.load_dataset("iris")

cols=["sepal_length","sepal_width","petal_length","petal_width"]
normalized_iris=iris.copy()

for i in cols:
  arr=np.array(iris[i])
  mean=np.mean(iris[i])
  std=np.std(iris[i])
  normalized_iris[i]=(arr-mean)/std
print("normalized data:",normalized_iris)
```
